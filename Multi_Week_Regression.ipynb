```python
from sklearn.neural_network import MLPClassifier

X = [[0., 0.], [1., 1.]]
y = [0, 1]
clf = MLPClassifier(solver='lbfgs', alpha=1e-5,
                    hidden_layer_sizes=(5, 2), random_state=1)

clf.fit(X, y)
clf.predict([[2., 2.], [-1., -2.]])
```

```python
[coef.shape for coef in clf.coefs_]
```

```python
clf.predict_proba([[2., 2.], [1., 2.]])
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
#sample is the number of points made, features is the number of attributes they have
#random_state controls the rng seed.
X, y = make_regression(n_samples=200, n_features=20, random_state=1)
#splits the points between training data and testing data. 75% of the points go to training, and the remaining 25% are held for tests
X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    random_state=1)
#This line creates the model, which is now stored in regr.
#max_iter determines how many iterations
#tol represents the minimum loss reduction to justify continuing(i.e. each step must match the equation better by this much)
#tldr: as tol decreases, the model is less likely to stop early.
regr = MLPRegressor(random_state=1, max_iter=10000, tol=0.0001)
regr.fit(X_train, y_train)

print(regr.predict(X_test[:2]))
print(y_test[:2])
regr.score(X_test, y_test)
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
import pandas as pd
import json
import numpy as np
df = pd.read_csv("roster.csv")

iterate_time = 227;
cur = 0;
X = []
y = []
while(cur < iterate_time):
  data = json.loads(df.iloc[cur]['stats'])

  records = []

  for period, stats in data.items():
    row = pd.json_normalize(stats, sep="_").iloc[0].to_dict()
    row['period'] = period
    records.append(row)

  df_player = pd.DataFrame(records)

  cols = ["period"] + [c for c in df_player.columns if c != "period"]
  df_player = df_player[cols]

  features = [
      "projected_points",
      "projected_avg_points",
      "projected_breakdown_receivingYards",
      "projected_breakdown_receivingReceptions",
      "projected_breakdown_receivingTouchdowns",
  ]
  goal = ["points"];

  df_features = df_player.reindex(columns=features)
  df_goal = df_player.reindex(columns = goal);
  any_empty = (df_features.iloc[0].isna().sum() + df_goal.iloc[0].isna().sum())
  if (any_empty == 0):
    X.append(df_features.iloc[0].to_numpy(dtype=float));
    y.append(df_goal.iloc[0]["points"])

  cur += 1

X = np.array(X, dtype=float)
y = np.array(y, dtype=float)
#splits the points between training data and testing data. 75% of the points go to training, and the remaining 25% are held for tests
X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    random_state=1)
#This line creates the model, which is now stored in regr.
#max_iter determines how many iterations
#tol represents the minimum loss reduction to justify continuing(i.e. each step must match the equation better by this much)
#tldr: as tol decreases, the model is less likely to stop early.
regr = MLPRegressor(random_state=1, max_iter=10000, tol=0.0001)
regr.fit(X_train, y_train)

print(regr.predict(X_test[:2]))
print(y_test[:2])
regr.score(X_test, y_test)
```
